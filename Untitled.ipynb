```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 6))

melted_sum = grouped_sum.melt(
    id_vars='Sub-Category',
    value_vars=['Profit', 'Sales'],
    var_name='Metric',
    value_name='Value'
)

sns.barplot(
    data=melted_sum,
    x='Sub-Category',
    y='Value',
    hue='Metric'
)

plt.title('Total Profit and Sales per Sub-Category')
plt.xlabel('Sub-Category')
plt.ylabel('Total Value')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
grouped_data = (
    df.groupby('Region')[['Sales', 'Profit']]
    .mean()
    .reset_index()
)

melted_data = grouped_data.melt(
    id_vars='Region',
    value_vars=['Sales', 'Profit'],
    var_name='Metric',
    value_name='Value'
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=melted_data,
    x='Region',
    y='Value',
    hue='Metric'
)

plt.title('Average Sales and Profit by Region')
plt.xlabel('Region')
plt.ylabel('Average Value')

plt.tight_layout()
plt.show()
```

```python
regions = df['Region'].unique()

fig, axes = plt.subplots(
    2, 2,
    figsize=(16, 12)
)

axes = axes.flatten()

for i, region in enumerate(regions):

    region_data = df[df['Region'] == region]

    state_data = (
        region_data
        .groupby('State')[['Sales', 'Profit']]
        .mean()
        .reset_index()
    )

    melted_region = state_data.melt(
        id_vars='State',
        value_vars=['Sales', 'Profit'],
        var_name='Metric',
        value_name='Value'
    )

    sns.barplot(
        data=melted_region,
        x='State',
        y='Value',
        hue='Metric',
        ax=axes[i]
    )

    axes[i].set_title(
        f'Average Sales and Profit in {region} Region'
    )

    axes[i].set_xlabel('State')
    axes[i].set_ylabel('Average Value')
    axes[i].tick_params(axis='x', rotation=90)

plt.tight_layout()
plt.show()
```

```python
discount_data = df_notime.sort_values('Discount')['Discount']

fig = go.Figure(
    data=[
        go.Histogram(
            x=discount_data
        )
    ]
)

fig.update_layout(
    title='Sales Histogram Plot for Each Discount',
    xaxis_title='Discount',
    yaxis_title='Number of Orders'
)

fig.show()
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

discount_values = sorted(df_notime['Discount'].unique())

fig = make_subplots(
    rows=6,
    cols=2,
    subplot_titles=[
        f'{int(d * 100)}% Discount'
        for d in discount_values
    ]
)

for i, discount in enumerate(discount_values):

    row = (i // 2) + 1
    col = (i % 2) + 1

    discount_sales = (
        df_notime[df_notime['Discount'] == discount]
        ['Sales']
    )

    fig.add_trace(
        go.Histogram(
            x=discount_sales,
            name=f'{int(discount * 100)}%'
        ),
        row=row,
        col=col
    )

fig.update_layout(
    height=1800,
    width=1000,
    title_text='Sales Distribution by Discount Level',
    showlegend=False
)

fig.show()
```

```python
df_time = df[
    [
        'Order Date',
        'Segment',
        'Sales',
        'Quantity',
        'Discount',
        'Profit'
    ]
].copy()

df_time['Year'] = df_time['Order Date'].dt.year
df_time['Month'] = df_time['Order Date'].dt.month
df_time['Month_Name'] = df_time['Order Date'].dt.month_name()

df_time.head()
```

```python
df_line = (
    df_time[
        ['Order Date', 'Sales', 'Profit']
    ]
    .sort_values('Order Date')
)

df_line.head()
```

```python
plt.figure(figsize=(16, 6))

plt.plot(
    df_line['Order Date'],
    df_line['Sales'],
    label='Sales'
)

plt.plot(
    df_line['Order Date'],
    df_line['Profit'],
    label='Profit'
)

plt.title('Sales and Profit Over Time')
plt.xlabel('Order Date')
plt.ylabel('Value')

plt.legend()
plt.tight_layout()
plt.show()
```

```python
monthly_sales = (
    df_time
    .groupby(['Year', 'Month'])['Sales']
    .sum()
    .reset_index()
)

monthly_sales.head()
```

```python
fig, axes = plt.subplots(
    3,
    1,
    figsize=(16, 12),
    sharey=True
)

axes = axes.flatten()

segment_list = [
    'Consumer',
    'Corporate',
    'Home Office'
]

for i, segment in enumerate(segment_list):

    segment_data = df_time[
        df_time['Segment'] == segment
    ]

    monthly_segment_sales = (
        segment_data
        .groupby(['Year', 'Month'])['Sales']
        .mean()
        .reset_index()
    )

    for year in sorted(
        monthly_segment_sales['Year'].unique()
    ):

        year_data = monthly_segment_sales[
            monthly_segment_sales['Year'] == year
        ]

        axes[i].plot(
            year_data['Month'],
            year_data['Sales'],
            marker='o',
            label=str(year)
        )

    axes[i].set_title(
        f'Average Sales for {segment}'
    )

    axes[i].set_xlabel('Month')
    axes[i].set_ylabel('Average Sales')

    axes[i].set_xticks(range(1, 13))
    axes[i].legend(title='Year')
    axes[i].grid(True)

plt.tight_layout()
plt.show()
```

```python
monthly_summary = (
    df_time
    .groupby('Month')['Sales']
    .sum()
    .sort_values(ascending=False)
)

print("Monthly Sales Ranking:")
print(monthly_summary)
```

```python
best_months = monthly_summary.head(4)
worst_months = monthly_summary.tail(4)

print("Best Sales Months:")
print(best_months)

print("\nLowest Sales Months:")
print(worst_months)
```
